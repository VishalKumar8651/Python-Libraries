```python
import numpy as np 
import pandas as pd
```

```python
data = pd.read_csv("dataset.csv")
data.head()
data.columns= data.columns.astype(int) #THE COLUMNS VALUES 0,1,2,3 ARE STORED IN STRING FORM SO WE CONVERTED IT
data = data.drop(columns=["Unnamed: 0.1"])
data = data.drop(columns=["Unnamed: 0"])
data.to_csv("dataset.csv",index=False)
```

```python
type(data)
data.columns = data.columns.astype(int) #Convert Column to Integer Type
data.columns
```

.loc[row_selection, column_selection]

```python
data.loc[[1,2],[0,1]] #1,2 rows of 0,1 columns
data.loc[[0,1,2],:]#ALL columns of 0,1,2 rows 
data.loc[:,[1]]#All rowws of 1st column
data.loc[:]#All Rows
data.loc[0,0] #Specific Value at 0th row and 0th column
# data.loc[(data.loc[:, 1] > 0.5)]
data.loc[(data[1] > 0.5, 1)] #Column 1 whose values are greater than 0.5
data.head(2)
```

iloc[] is used when we want values based index of rows and columns (row/column name could be "A", "State"...)

```python
data.iloc[[0],[1]]
```

Data.drop() wont change the actual DataFrame it just return the copy or view 
TO actually change the dataset we have two options :
    1. We Can write :- data = data.drop(1) ----- Using Assignment Operator
    2. We can Use "inplace = True" ---- This will automatically reflect the changes in actual dataset
    

```python
data.drop(0)
data.drop(columns=1)
data.head()
data = data.drop(1)
data.head()
# here we can see the actual data is modified and 3 is removed but not 0 
data.drop(2,inplace=True)
data.head()
#Now 2nd row is also removed from the dataset
```

```python
data.drop(1,axis=0)
data.drop(1,axis=1)
```

```python
data.reset_index()
data.reset_index(drop=True) #Removes the extra index column
```

```python
data.dropna() #removes all Null Values written as 'pd.NaT'
data.dropna(how='all',axis=1) #Removes If all avlues in axis is null
data.isnull()
data.notnull()
```
