```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv("Practice.csv")
```

```python
#Find all permits from zipcode 94102
df[df["Zipcode"]==94102]
```

```python
#Find rows where street number > 1000

df[df["Street Number"]>1000]
```

```python
#Find all demolition permits
df[df["Permit Type Definition"]=="demolitions"]
```

```python
#Find rows having missing zipcode

df[df["Zipcode"].isna()]
```

```python
#Find permits from specific neighborhoods

df[df["Neighborhoods - Analysis Boundaries"]=="Tenderloin"]
```

```python
#Sort by street number descending

df["Street Number"].sort_values(ascending=False)
```

```python
#Sort by creation date
df["Permit Creation Date"].sort_values()
```

```python
#Top zipcodes by number of permits

df["Number of Existing Stories"].sort_values(ascending=False)
```

```python
#Multi-column sorting


df.sort_values(["Street Number" ,"Zipcode"], ascending=[False,True])
```

```python
#Rows with more than 500 street number and Zipcode = 94102

df.loc[(df["Street Number"]>500) & (df["Zipcode"]==94102)]
```
