6. Aggregation
Questions:
Max street number
Mean construction type

Multiple aggregations together

```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv("Practice.csv")
df.dropna(how="all")
df
```

```python
df["Zipcode"].unique()
```

```python
#Count permits per zipcode
df.groupby("Zipcode")["Permit Number"].count()
```

```python
#Average street number per neighborhood
df.groupby("Neighborhoods - Analysis Boundaries")["Street Number"].mean()
```

```python
#Which permit type occurs most?
df["Permit Type"].idxmax()
```

```python
#Count permits by supervisor district

df.groupby("Supervisor District")["Permit Number"].count()
```

```python
#Group by permit type + zipcode

df.groupby(["Zipcode" , "Permit Type"]).count()
```
