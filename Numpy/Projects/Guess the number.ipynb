Guess the random number

```python
import numpy as np
import random
```

```python
x= random.randint(0,100)

guess = int(input("Guess the number"))
    
while(True):
    if(guess<x):
        print("More High")
        guess = int(input("Guess Again"))
        
    if(guess>x):
        print("More Less")
        guess = int(input("Guess Again"))
       
    if(guess==x):
        print("You guesed the number ", x)
        break;
```
