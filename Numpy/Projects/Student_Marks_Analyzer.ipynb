```python
# Rows = students, Columns = subjects
import numpy as np

marks = np.array([
    [72, 85, 90, 68],
    [88, 79, 95, 92],
    [60, 70, 65, 58],
    [91, 89, 94, 99],
    [76, 84, 80, 73]
])
```

Average  per student

Average  per subject

Highest /lowest scores

Standard  deviation

Rank  students

Normalize  marks


```python
avg = []

avg = np.mean(marks, axis=1)
print("Average Per Student is :")
for i in range(avg.shape[0]):
    print(i, avg[i])
```

```python
avg = []

avg = np.mean(marks, axis=0)
print("Average Per Subject is :")
for i in range(avg.shape[0]):
    print(i, avg[i])
```

```python
print(np.max(marks))
print(np.min(marks))
```

```python
rank = []
rank = np.sort(avg)

for i in range(avg.shape[0]):
    print(i ,rank[i])
```

```python
print("Standard Deviation :", np.std(marks,axis=1))
```

```python
#NORMALIZATION
print("MAX-MIN METHOD NORMALIZATIO\n")
norm = (marks - marks.min())/(marks.max()-marks.min())
print(norm)

# .........
print("Z-SCORE NORMALIZATION\n")
norm = (marks - np.mean(marks)) / np.std(marks)
print(norm)
```
