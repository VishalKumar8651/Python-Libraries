```python
import numpy as np
rows1 = int(input("Enter number of rows for Matrix 1: "))
cols1 = int(input("Enter number of columns for Matrix 1: "))

print("Enter elements of Matrix 1:")
matrix1 = []
for i in range(rows1):
    row = list(map(int, input().split()))
    matrix1.append(row)

matrix1 = np.array(matrix1)


# Input for second matrix
rows2 = int(input("Enter number of rows for Matrix 2: "))
cols2 = int(input("Enter number of columns for Matrix 2: "))

print("Enter elements of Matrix 2:\n")
matrix2 = []
for i in range(rows2):
    row = list(map(int, input().split()))
    matrix2.append(row)

matrix2 = np.array(matrix2)



operation = input("Enter the Operation")

def resulting_func(matrix1 ,matrix2):
    if(operation =="+"):
        final_matrix = matrix1+matrix2
    elif(operation =="-"): 
        final_matrix = matrix1-matrix2
    elif(operation =="*"):
        final_matrix = np.dot(matrix1,matrix2)
    elif(operation =="T"):
        final_matrix = matrix1.T
    elif(operation =="/"):
        if(np.any(matrix2==0)):
            return 0
        else:
            final_matrix = matrix1/matrix2

    else:
            return "Invalid Operation"

    return final_matrix

    

ans = resulting_func(matrix1, matrix2)
print("\n Result :")
print(ans)
```
